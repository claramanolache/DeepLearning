```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import keras
import torch
import torch.nn as nn
from torch.utils import data

from sklearn import model_selection
from sklearn import ensemble
from sklearn import metrics
from sklearn import dummy
```

```python
if torch.cuda.is_available():
    device = "cuda"
elif torch.backends.mps.is_available():
    device = "mps"
else:
    device = "cpu"
```

```python
seed = 42
np.random.seed(seed)
torch.manual_seed(seed)
```

##Helper Functions

```python
def init_params(module):
    if isinstance(module, nn.Linear):
        nn.init.kaiming_uniform_(module.weight)
        nn.init.zeros_(module.bias)
```

```python
def reconstruction_error(X):
    model.eval()

    with torch.no_grad():
        X = X.to(device)
        reconstructed = model(X)

        errors = torch.mean(
            (X - reconstructed) ** 2,
            dim=1
        )

    return errors.cpu().numpy()
```

going to encode in ranges

```python
common_ports = (
    X["Destination Port"]
    .value_counts()
    .head(12)
    .index.tolist()
)
```

```python
def categorize_port(port, common_ports):
    if port in common_ports:
        return str(port)
    elif port == 0:
        return "reserved"
    elif port <= 1023:
        return "well_known"
    elif port <= 49151:
        return "registered"
    else:
        return "dynamic"
```

```python
X["Destination Port"] = X["Destination Port"].apply(
    lambda port: categorize_port(port, common_ports)
)
```

```python
X = pd.get_dummies(
    X,
    columns=["Destination Port"],
    dtype=float
)
```

###Encoding source ports

```python
df['Source Port'].nunique()
```

```python
common_source_ports = (
    X["Source Port"]
    .value_counts()
    .head(12)
    .index.tolist()
)

X["Source Port"] = X["Source Port"].apply(
    lambda port: categorize_port(port, common_source_ports)
)

X = pd.get_dummies(
    X,
    columns=["Source Port"],
    dtype=float
)

X.head()
```

###Encode NAF

```python
X["NAT Source Changed"] = (
    df["Source Port"] != df["NAT Source Port"]
).astype(int)

X["NAT Destination Changed"] = (
    df["Destination Port"] != df["NAT Destination Port"]
).astype(int)
```

```python
common_nat_ports = (
    X["NAT Destination Port"]
    .value_counts()
    .head(12)
    .index.tolist()
)

X["NAT Destination Port"] = X["NAT Destination Port"].apply(
    lambda port: categorize_port(port, common_nat_ports)
)

X["NAT Source Port"] = X["NAT Source Port"].apply(
    lambda port: categorize_port(port, [])
)

X = pd.get_dummies(
    X,
    columns=["NAT Source Port", "NAT Destination Port"],
    dtype=float
)
```

###rest of splitting

```python
import pandas as pd
from sklearn.model_selection import train_test_split

y_labels = (actions != 'allow').astype(int)

X_sub = X.copy()
y_sub = y_labels.copy()

normal_indices = y_sub[y_sub == 0].index
anomaly_indices = y_sub[y_sub == 1].index

train_normal_idx, rem_normal_idx = train_test_split(
    normal_indices,
    test_size=0.40,
    random_state=seed
)

X_train = X_sub.loc[train_normal_idx]
y_train = y_sub.loc[train_normal_idx]

mixed_rem_idx = rem_normal_idx.union(anomaly_indices)
X_rem = X_sub.loc[mixed_rem_idx]
y_rem = y_sub.loc[mixed_rem_idx]

X_val, X_test, y_val, y_test = train_test_split(
    X_rem, y_rem,
    test_size=0.5,
    random_state=seed,
    stratify=y_rem
)

print(f"Training set (Normal only): {X_train.shape}, {y_train.value_counts().to_dict()}")
print(f"Validation set (Mixed): {X_val.shape}, {y_val.value_counts().to_dict()}")
print(f"Test set (Mixed): {X_test.shape}, {y_test.value_counts().to_dict()}")
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

X_train = scaler.fit_transform(X_train)
X_val = scaler.transform(X_val)
X_test = scaler.transform(X_test)
```

```python
train_loader = data.DataLoader(
    data.TensorDataset(
        torch.FloatTensor(X_train)
    ),
    batch_size=32,
    shuffle=True
)

X_val = torch.FloatTensor(X_val)
X_test = torch.FloatTensor(X_test)
```

##Autencoder/Decoder shapes

```python
input_dim = X_train.shape[1]
print(input_dim)
```

```python
latent_dim = 4

encoder = nn.Sequential(
    nn.Linear(input_dim, 32),
    nn.ReLU(),
    nn.Linear(32, 16),
    nn.ReLU(),
    nn.Linear(16, latent_dim)
)
```

```python
decoder = nn.Sequential(
    nn.Linear(latent_dim, 16),
    nn.ReLU(),
    nn.Linear(16, 32),
    nn.ReLU(),
    nn.Linear(32, input_dim)
)
```

```python
model = nn.Sequential(
    encoder,
    decoder
).apply(init_params)

model = model.to(device)
```

##Hyperparams

```python
n_epochs = 30
learning_rate = 0.001

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=learning_rate
)

criterion = nn.MSELoss()
```

##Training

```python
X_val_normal = X_val[(y_val == 0).values]

train_losses = []
val_losses = []

for epoch in range(n_epochs):

    model.train()
    total_loss = 0

    for (X_batch,) in train_loader:

        X_batch = X_batch.to(device)

        optimizer.zero_grad()

        reconstructed = model(X_batch)
        loss = criterion(reconstructed, X_batch)

        loss.backward()
        optimizer.step()

        total_loss += loss.item() * X_batch.size(0)

    train_loss = total_loss / len(X_train)
    train_losses.append(train_loss)

    model.eval()

    with torch.no_grad():
        val_batch = X_val_normal.to(device)
        reconstructed_val = model(val_batch)

        val_loss = criterion(
            reconstructed_val,
            val_batch
        ).item()

    val_losses.append(val_loss)

    print(
        f"Epoch {epoch+1}/{n_epochs} | "
        f"Train: {train_loss:.6f} | "
        f"Val: {val_loss:.6f}"
    )
```

##Results/graphs

```python
plt.plot(train_losses, label="Training Loss")
plt.plot(val_losses, label="Validation Loss")

plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.title("Autoencoder Training")
plt.legend()
plt.show()
```

```python
val_errors = reconstruction_error(X_val_normal)

threshold = np.percentile(val_errors, 95)

print("Anomaly threshold:", threshold)
```

```python
test_errors = reconstruction_error(X_test)

y_pred = (test_errors > threshold).astype(int)
```

```python
print(metrics.classification_report(
    y_test,
    y_pred,
    target_names=["Allow", "Non-Allow"]
))
```

```python
metrics.ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    display_labels=["Allow", "Non-Allow"]
)

plt.show()
```

```python
auc = metrics.roc_auc_score(y_test, test_errors)
print("ROC-AUC:", auc)
```
